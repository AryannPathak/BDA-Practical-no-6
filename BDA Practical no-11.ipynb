```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder \
    .appName("Spark SQL Analysis") \
    .getOrCreate()

print("Spark Session Created Successfully")
```

```python
data = [
    (1, "Amit", "AIML", 20, 85),
    (2, "Sneha", "CSE", 21, 90),
    (3, "Rahul", "AIML", 20, 75),
    (4, "Priya", "IT", 22, 88),
    (5, "Neha", "CSE", 21, 95),
    (6, "Rohit", "IT", 20, 65),
    (7, "Pooja", "AIML", 21, 80),
    (8, "Kiran", "CSE", 22, 70)
]

columns = ["ID", "Name", "Department", "Age", "Marks"]

df = spark.createDataFrame(data, columns)

df.show()
```

```python
df.createOrReplaceTempView("students")

print("Temporary View Created Successfully")
```

```python
spark.sql("SELECT * FROM students").show()
```

```python
spark.sql("""
SELECT Name, Department, Marks
FROM students
WHERE Marks > 80
""").show()
```

```python
spark.sql("""
SELECT *
FROM students
WHERE Department = 'AIML'
""").show()
```

```python
spark.sql("""
SELECT Name, Marks
FROM students
ORDER BY Marks DESC
""").show()
```

```python
spark.sql("""
SELECT COUNT(*) AS Total_Students
FROM students
""").show()
```

```python
spark.sql("""
SELECT
    AVG(Marks) AS Average_Marks,
    MAX(Marks) AS Maximum_Marks,
    MIN(Marks) AS Minimum_Marks
FROM students
""").show()
```

```python
spark.sql("""
SELECT
    Department,
    COUNT(*) AS Total_Students
FROM students
GROUP BY Department
""").show()
```

```python
spark.sql("""
SELECT
    Department,
    AVG(Marks) AS Average_Marks
FROM students
GROUP BY Department
""").show()
```
