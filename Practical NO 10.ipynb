```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

```python
spark = SparkSession.builder \
    .appName("CSV Data Analysis") \
    .getOrCreate()

print("Spark Session Created Successfully")
```

```python
data = [
    (1, "Amit", "AIML", 20, 85),
    (2, "Sneha", "CSE", 21, 90),
    (3, "Rahul", "AIML", 20, 75),
    (4, "Priya", "IT", 22, 88),
    (5, "Neha", "CSE", 21, 95),
    (6, "Rohit", "IT", 20, 65),
    (7, "Pooja", "AIML", 21, 80),
    (8, "Kiran", "CSE", 22, 70)
]

columns = ["ID", "Name", "Department", "Age", "Marks"]

df = spark.createDataFrame(data, columns)

df.write.mode("overwrite").option("header", True).csv("student_data")
```

```python
df = spark.read.csv(
    "student_data",
    header=True,
    inferSchema=True
)

print("Student Dataset:")
df.show()
```

```python
df.printSchema()
```

```python
df.select("Name", "Department", "Marks").show()
```

```python
df.filter(df.Marks > 80).show()
```

```python
df = df.withColumn(
    "Result",
    when(df.Marks >= 40, "Pass").otherwise("Fail")
)

df.show()
```

```python
df.orderBy(df.Marks.desc()).show()
```

```python
df.describe("Age", "Marks").show()
```

```python
print("Total Students:", df.count())
```
